## Visualizacion de datos

```python
import plotly.express as px
```

```python
px.histogram(data, x="tienda")
```

```python
grafico =px.histogram(data,
             x="tienda",
             text_auto=True, 
             title="Facturacion",
             color="forma_pago")
grafico.show()
grafico.write_html("Facturacion.html")
```

```python
columnas = ["tienda", "Ciudad", "País", "tamaño", "local_consumo"]

for columna in columnas:
    fig = px.histogram(data,
        x=columna,
        y='precio',
        color='forma_pago',
        text_auto=True,
        title="Facturacion"
    )
    fig.write_html(f"Facturacion por {columna}.html")
    fig.show()
```

```python
agrupado = data.groupby(["tienda", "fecha"]).precio.sum().to_frame()

agrupado.reset_index(inplace=True)

agrupado["acumulado"]= agrupado.groupby("tienda").precio.cumsum()
```

```python
fig = px.bar(agrupado,
             x='acumulado',
             y='tienda',
             color='tienda',
             text_auto=True,
             range_x=[0, 100000],
             animation_frame="fecha")
fig.show()
```
